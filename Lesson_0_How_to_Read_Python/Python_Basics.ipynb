    <div style="font-weight:700;font-size:1.08em;margin-bottom:4px">
        💡 The rules for names
    </div>
    A <b>name</b> (Python calls it an identifier) is what you give a value so you can use it later.
    <ul>
        <li>
            It may contain <b>letters, digits and underscores</b>. It may <b>not start with a digit</b>, and may not contain spaces or punctuation such as <code>-</code>, <code>@</code>, <code>$</code> or <code>%</code>.
        </li>
        <li>
            Names are <b>case sensitive</b>: <code>Site</code> and <code>site</code> are two different names.
        </li>
        <li>
            About thirty <b>reserved words</b> (<code>for</code>, <code>if</code>, <code>and</code>, <code>def</code>, <code>import</code>, ...) belong to the language and cannot be names.
        </li>
        <li>
            Avoid names that Python already uses for tools, such as <code>list</code>, <code>str</code>, <code>max</code> and <code>sum</code>. Python lets you reuse them, and your own copy then hides the original until you restart.
        </li>
    </ul>
</div>

<div style="border-left:7px solid #2f80ed;background:rgba(47,128,237,.12);padding:10px 18px;margin:10px 0;border-radius:8px;line-height:1.5">
    <div style="font-weight:700;font-size:1.08em;margin-bottom:4px">
        💡 Slicing: <code>[start:stop:step]</code>
    </div>
    <ul>
        <li>
            <b>start</b> is the first item you want (default: the beginning).
        </li>
        <li>
            <b>stop</b> is the first item you do <b>not</b> want (default: the end).
        </li>
        <li>
            <b>step</b> is how far to move each time (default: 1).
        </li>
        <li>
            <b>Negative numbers count from the end</b>: <code>-1</code> is the last item.
        </li>
        <li>
            <code>+</code> joins strings, <code>*</code> repeats them, <code>len()</code> counts the characters.
        </li>
    </ul>
    The same rules work for lists, tuples and (later) table rows.
</div>

```python
# WORKED EXAMPLE: an AQS-style site ID, written state-county-site
site_id = "24-033-8003"

print(site_id)
print(site_id[0])       # first character
print(site_id[:2])      # the first two
print(site_id[-4:])     # the last four
print(len(site_id))
print(site_id[::-1])    # a negative step goes backwards
```

```python
# WORKED EXAMPLE: run it, then compare with your prediction
readings = {"A": [38, 52, 61, 74], "B": [22, 35, 41, 30], "C": [12, 18, 15, 20]}
limit = 40
report = {}

for site, values in readings.items():
    n_over = 0
    for x in values:
        if x > limit:
            n_over += 1
    report[site] = n_over
print(report)
reveal("script", {1: "A", 4: "B"}.get(report["B"], "C"))
```
