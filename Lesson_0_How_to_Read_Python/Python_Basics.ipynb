<div style="border-left:7px solid #2f80ed;background:rgba(47,128,237,.12);padding:10px 18px;margin:10px 0;border-radius:8px;line-height:1.5">
    How does Python know that <code>7</code> is a number? Nobody told it. It looked at how the thing was written.
    <p>
        Suppose I write <code>7</code>. Bare digits: Python takes it for a whole number. Now suppose I write <code>7.0</code>. That is the same quantity, but the decimal point is a signal, and Python files it as a <b>float</b>, a different sort of number, stored in a different way. Now put quotation marks around the digit, <code>"7"</code>, and something odd happens. It stops being a number. It is a character, the sort of thing you would find in a name or a sentence.
    </p>
</div>

<div style="border-left:7px solid #2f80ed;background:rgba(47,128,237,.12);padding:10px 18px;margin:10px 0;border-radius:8px;line-height:1.5">
    Suppose I run <code>print(7)</code>, and then <code>print("7")</code>. Both put a 7 on the screen. Nothing you can see tells them apart, yet one is a number and the other is a character, and the difference shows the moment you try to do arithmetic with them. So how do you find out which one you are holding?
    <p>
        You ask Python. Run <code>type("7")</code> and it answers <code>&lt;class 'str'&gt;</code>. Python was never confused, since the quotes told it everything, and it will say so if you ask. What I typed, a name followed by parentheses with a value inside, is a <b>function call</b>: I handed <code>type</code> a value, and it handed back its answer.
    </p>
</div>

```python
print(7)
print("7")
print(type(7), type(7.0), type("7"))
```

```python
# one value of each kind from the box above
print(type(7))
print(type(41.2))
print(type("Site A"))
print(type(True))
print(type([38, 52, 61]))
print(type((60264, 28)))
print(type({"A": 41.2}))
```

<div style="border-left:7px solid #2f80ed;background:rgba(47,128,237,.12);padding:10px 18px;margin:10px 0;border-radius:8px;line-height:1.5">
    What happens if you ask Python to add a number to a piece of text? Suppose I write <code>365 + "365"</code>. You might expect 730, or you might expect the text <code>"365365"</code>, and a language could reasonably do either. Python does neither. It stops and prints a red message, and the line to read is the last one: <code>TypeError: unsupported operand type(s) for +: 'int' and 'str'</code>. Python calls this a <b>TypeError</b>: the types do not fit the operation. The lines above it only tell you where.
</div>

```python
# This fails on purpose. Run it, then read the red message from the bottom up.
365 + "365"
```

<div style="border-left:7px solid #2f80ed;background:rgba(47,128,237,.12);padding:10px 18px;margin:10px 0;border-radius:8px;line-height:1.5">
    How do you turn the text <code>"41.2"</code> into a number? You hand it to <code>float()</code>, and out comes <code>41.2</code>. <code>int()</code> does the same for whole numbers, so <code>int("41")</code> gives 41. But try <code>int("41.2")</code> and Python refuses. The text has a decimal point in it, and <code>int()</code> will not guess that you meant to throw the <code>.2</code> away. Either way, the original is left alone and you get a new value back.
</div>

```python
print(float("41.2"))
print(int("41"))

# This one fails on purpose.
int("41.2")
```

<div style="border-left:7px solid #2f80ed;background:rgba(47,128,237,.12);padding:10px 18px;margin:10px 0;border-radius:8px;line-height:1.5">
    What does <code>int(7.9)</code> give back? Not 8. It gives 7. <code>int()</code> does not round; it cuts off whatever follows the decimal point, and for 7.9 the cut leaves 7. If you want 8, ask for it with <code>round(7.9)</code>. Both give back a whole number, which is why the difference is easy to forget, and only one of them looked at the <code>.9</code>.
</div>

```python
print(int(7.9), round(7.9))
```

<div style="border-left:7px solid #2f80ed;background:rgba(47,128,237,.12);padding:10px 18px;margin:10px 0;border-radius:8px;line-height:1.5">
    What are you allowed to call a value? Suppose I try <code>2nd_site = 1</code>. Python refuses, because it starts reading <code>2nd_site</code> as the number 2 and then does not know what to do with the rest. So a name cannot start with a digit. After that, letters, digits and underscores are fine, and not much else is: <code>site-a</code> would be read as <code>site</code> minus <code>a</code>. Capitals count, so <code>Site</code> and <code>site</code> are two different names. Words Python keeps for itself, such as <code>for</code> and <code>if</code>, are taken. Avoid <code>list</code>, <code>str</code> and <code>sum</code> too: Python lets you reuse them, and your version then hides the original.
</div>

```python
site_2 = 41.2
print(site_2)

# Remove the # in front of one line to see it fail, then put it back.
# 2nd_site = 1
# site-a = 1
```

<div style="border-left:7px solid #2f80ed;background:rgba(47,128,237,.12);padding:10px 18px;margin:10px 0;border-radius:8px;line-height:1.5">
    Is <code>x = x + 1</code> nonsense? As an equation it is: no number equals itself plus one. But Python does not read <code>=</code> as an equation. It reads it as an instruction: work out the right side, then store the result under the name on the left. So I write <code>x = 5</code>, and then <code>x = x + 1</code>. The right side comes to 6, and 6 goes into <code>x</code>. This is called <b>assignment</b>. Nothing had to announce <code>x</code> beforehand; the first assignment brings it into being.
</div>

```python
x = 5
x = x + 1
print(x)

# Several names in one line. The right side is worked out first, so this swaps them.
low, high = 70, 35
low, high = high, low
print(low, high)
```

<div style="border-left:7px solid #2f80ed;background:rgba(47,128,237,.12);padding:10px 18px;margin:10px 0;border-radius:8px;line-height:1.5">
    How does Python know where a group of instructions ends? Look at the code below, without worrying yet about what <code>if</code> does. A line that ends in a colon opens a block, and the lines pushed in underneath it belong to that block. When the pushing in stops, the block is over. Other languages use braces for this. Python uses the layout itself, four spaces by convention, and every line in a block must be pushed in by the same amount. Move one line and you have changed the program.
</div>

```python
a = 5.0
if a >= 0:
    print("Positive")
else:
    print("Negative")
    print("(this line is part of else)")
```

<div style="border-left:7px solid #2f80ed;background:rgba(47,128,237,.12);padding:10px 18px;margin:10px 0;border-radius:8px;line-height:1.5">
    How do you pull part of a piece of text out of the whole? Take the site ID <code>"24-033-8003"</code>, which reads state, county, site. Say I want the state, the first two characters. Python counts positions from 0, so those are positions 0 and 1, and I write <code>site_id[0:2]</code>. The second number is the first position I do not want, which is why it is 2 and not 1. That pair of numbers in brackets is a <b>slice</b>. Leave out the first and you start at the beginning; leave out the second and you run to the end. A negative number counts from the end, so <code>site_id[-4:]</code> is the last four characters, <code>"8003"</code>.
</div>

```python
site_id = "24-033-8003"

print(site_id[0])       # one character
print(site_id[:2])      # the state
print(site_id[3:6])     # the county
print(site_id[-4:])     # the last four
print(len(site_id))     # how many characters
print(site_id[::-1])    # a third number is a step; a negative step goes backwards
```

<div style="border-left:7px solid #2f80ed;background:rgba(47,128,237,.12);padding:10px 18px;margin:10px 0;border-radius:8px;line-height:1.5">
    What if you have six readings rather than one? Put them in square brackets with commas between, and give the whole thing a name: <code>readings = [38, 52, 61, 74, 69, 45]</code>. Everything you just learned about positions carries over, because a list is read the way text is. The first reading is <code>readings[0]</code>, the last is <code>readings[-1]</code>, and <code>readings[1:3]</code> gives the second and third. The difference is that a list can be changed. I can write <code>readings[0] = 40</code> and the first reading becomes 40, or <code>readings.append(50)</code> and a seventh reading appears at the end.
</div>

```python
readings = [38, 52, 61, 74, 69, 45]

print(readings[0], readings[-1])
print(readings[1:3])
readings[0] = 40
readings.append(50)
print(readings)
print(len(readings))
```

<div style="border-left:7px solid #2f80ed;background:rgba(47,128,237,.12);padding:10px 18px;margin:10px 0;border-radius:8px;line-height:1.5">
    If I write <code>copy_of = readings</code>, do I have two lists? No. I have one list with two names. Change it through either name and the other name sees the change: after <code>copy_of[0] = 99</code>, <code>readings[0]</code> is 99 as well. If you want a list that is really separate, ask for one with <code>readings.copy()</code>.
</div>

```python
readings = [38, 52, 61]
copy_of = readings
copy_of[0] = 99
print(readings[0])      # 99: one list, two names

separate = readings.copy()
separate[0] = 0
print(readings[0])      # still 99: the copy is its own list
```

<div style="border-left:7px solid #2f80ed;background:rgba(47,128,237,.12);padding:10px 18px;margin:10px 0;border-radius:8px;line-height:1.5">
    Why would you want a list you are not allowed to change? Take the shape of a table, its number of rows and columns: <code>(60264, 28)</code>. Those two numbers belong together, and nothing you do afterwards should quietly turn the 28 into a 30. Reading works as it does for a list, so <code>shape[1]</code> is 28. Writing does not. <code>shape[0] = 100</code> stops with a <code>TypeError</code>, and the message says a tuple does not support item assignment.
</div>

```python
shape = (60264, 28)
print(shape[1])

v = 1, 2, 3             # commas alone make a tuple; the parentheses are optional
print(type(v))

# This one fails on purpose.
shape[0] = 100
```

<div style="border-left:7px solid #2f80ed;background:rgba(47,128,237,.12);padding:10px 18px;margin:10px 0;border-radius:8px;line-height:1.5">
    How do you look something up by its name instead of its position? Suppose I keep the months of each season, each under the season's name: <code>season = {"Spring": [3, 4, 5], "Fall": [9, 10, 11]}</code>. I write <code>season["Fall"]</code> and get back <code>[9, 10, 11]</code>. The name in the brackets is the <b>key</b>, and what comes back is its value. There is no counting of positions, and the order I wrote the pairs in does not matter to the lookup.
</div>

```python
season = {"Spring": [3, 4, 5], "Fall": [9, 10, 11]}

print(season["Fall"])
season["Summer"] = [6, 7, 8]    # assign to a key that is not there yet to add a pair
print(season.keys())
print(season.values())
print(season.items())           # (key, value) pairs
print(len(season))
```

<div style="border-left:7px solid #2f80ed;background:rgba(47,128,237,.12);padding:10px 18px;margin:10px 0;border-radius:8px;line-height:1.5">
    How many whole days are in 190 hours, and how many hours are left over? Plain division gives <code>190 / 24</code>, which is 7.9166..., and that is neither number you wanted. Two other operators split it the way you would on paper. <code>190 // 24</code> asks how many times 24 fits, and the answer is 7. <code>190 % 24</code> asks what remains once those 7 days are taken out: 7 days is 168 hours, and 190 less 168 is 22.
</div>

```python
print(190 / 24)
print(190 // 24)
print(190 % 24)

print(7 + 3, 7 - 3, 7 * 3)      # these three are as you expect
print(3 ** 2)                   # ** is the power
```

<div style="border-left:7px solid #2f80ed;background:rgba(47,128,237,.12);padding:10px 18px;margin:10px 0;border-radius:8px;line-height:1.5">
    What if you have a list of readings and want the largest, the smallest and the total? Python already has functions waiting for that. With <code>readings = [38, 52, 61, 74, 69, 45]</code>, <code>max(readings)</code> is 74, <code>min(readings)</code> is 38, <code>sum(readings)</code> is 339, and <code>len(readings)</code> is 6. The average is the total divided by the count, which is 56.5, and <code>round(avg, 1)</code> keeps one decimal place.
</div>

```python
readings = [38, 52, 61, 74, 69, 45]

print(max(readings), min(readings))
print(sum(readings), len(readings))

avg = sum(readings) / len(readings)
print(round(avg, 1))
```

<div style="border-left:7px solid #2f80ed;background:rgba(47,128,237,.12);padding:10px 18px;margin:10px 0;border-radius:8px;line-height:1.5">
    Where is the square root? It is not among the operators, and it is not built in either. Python keeps the less common mathematics in a toolbox called a <b>module</b>, and you ask for the toolbox by name before you use it: <code>import math</code>. After that, <code>math.sqrt(16)</code> gives 4.0, and the dot means "the <code>sqrt</code> that lives inside <code>math</code>". <code>math.log10(1000)</code> gives 3.0, because 10 raised to the power 3 is 1000.
</div>

```python
import math

print(math.sqrt(16))
print(math.log10(1000))
print(math.pi)
```

<div style="border-left:7px solid #2f80ed;background:rgba(47,128,237,.12);padding:10px 18px;margin:10px 0;border-radius:8px;line-height:1.5">
    What is the difference between <code>x = 5</code> and <code>x == 5</code>? The first stores: it puts 5 under the name <code>x</code>. The second asks: it works out whether <code>x</code> is 5, and the answer is itself a value, <code>True</code> or <code>False</code>. The other comparisons, <code>!=</code>, <code>&gt;</code>, <code>&lt;</code>, <code>&gt;=</code> and <code>&lt;=</code>, ask in the same way. Swap the one equals sign for the two, or the reverse, and on a line of its own nothing complains.
</div>

```python
x = 5
print(x == 5)

a, b = 21, 10
print(a == b)
print(a != b)
print(a > b)
print(a <= b)
```

<div style="border-left:7px solid #2f80ed;background:rgba(47,128,237,.12);padding:10px 18px;margin:10px 0;border-radius:8px;line-height:1.5">
    Are <code>"35"</code> and <code>35</code> equal? Python says no, and it is right to be strict. Consider the county code <code>"033"</code>. It is a label, not a quantity. Turn it into a number and you get 33, the same number you get from the different label <code>"33"</code>, and the zero that told them apart is gone. If you do want text and number to match, you convert one side, on purpose.
</div>

```python
print("35" == 35)
print("033" == "33")                 # two different labels
print(int("033"), int("33"))        # one number
print(int("35") == 35)              # converted on purpose
```

<div style="border-left:7px solid #2f80ed;background:rgba(47,128,237,.12);padding:10px 18px;margin:10px 0;border-radius:8px;line-height:1.5">
    How do you say that a reading is plausible if it lies between 0 and 500? You write both halves and join them with <code>and</code>: <code>reading &gt;= 0 and reading &lt;= 500</code>. The whole is True only when both halves are. <code>or</code> is True when at least one half is, and <code>not</code> flips an answer. Try 41.2, and then -999.0, a common code for a missing reading. The second comes out False.
</div>

```python
reading_a = 41.2
reading_b = -999.0

print(reading_a >= 0 and reading_a <= 500)
print(reading_b >= 0 and reading_b <= 500)

print(False or True)
print(not True)
```

<div style="border-left:7px solid #2f80ed;background:rgba(47,128,237,.12);padding:10px 18px;margin:10px 0;border-radius:8px;line-height:1.5">
    <div style="font-weight:700;font-size:1.08em;margin-bottom:4px">
        💡 How to read an <code>if</code>
    </div>
    <ul>
        <li>
            <code>if condition:</code> runs the indented lines <b>only when the condition is True</b>.
        </li>
        <li>
            <code>elif condition:</code> (“else if”) is tried only if everything above it was False.
        </li>
        <li>
            <code>else:</code> runs when nothing above it did.
        </li>
        <li>
            Python goes <b>top to bottom and stops at the first True</b>. Order matters.
        </li>
        <li>
            Zero, an empty string and an empty list count as False; nearly everything else counts as True.
        </li>
    </ul>
</div>

<div style="border-left:7px solid #2f80ed;background:rgba(47,128,237,.12);padding:10px 18px;margin:10px 0;border-radius:8px;line-height:1.5">
    How does Python choose between three outcomes? Take a reading of 41.2 and three invented bands, which are not regulatory categories: below 12 is low, below 35 is moderate, anything else is high. Python tests <code>pm25 &lt; 12</code> first. It is False, so Python moves on to <code>pm25 &lt; 35</code>, which is also False, and falls to <code>else</code>. Of the three indented lines, only one runs.
</div>

```python
pm25 = 41.2

if pm25 < 12:
    band = "low"
elif pm25 < 35:
    band = "moderate"
else:
    band = "high"

print(pm25, "->", band)
```

<div style="border-left:7px solid #2f80ed;background:rgba(47,128,237,.12);padding:10px 18px;margin:10px 0;border-radius:8px;line-height:1.5">
    Which band is a reading of exactly 35.0 in? Follow the tests. Is 35.0 below 12? No. Is it below 35? No, because 35.0 is not below itself, so that test is False too and the code falls through to <code>else</code>. The band is high. Had I written <code>&lt;=</code>, it would be moderate. Which is right depends on how the rule you are following is worded, and the comparison operator is where the code records that decision.
</div>

```python
pm25 = 35.0

if pm25 < 12:
    band = "low"
elif pm25 < 35:
    band = "moderate"
else:
    band = "high"

print(band)
print(pm25 < 35, pm25 <= 35)
```

<div style="border-left:7px solid #2f80ed;background:rgba(47,128,237,.12);padding:10px 18px;margin:10px 0;border-radius:8px;line-height:1.5">
    Can one decision sit inside another? It can. The inner lines are pushed in one level further, and Python reaches them only when the outer test passes. Take a PM2.5 of 41.2 with the wind at 3 mph. The outer test, above 35, is True, so Python goes in and asks the inner question: is the wind below 5?
</div>

```python
pm25, wind_mph = 41.2, 3

if pm25 > 35:
    if wind_mph < 5:
        msg = "high PM2.5 and calm air: pollution may be building up"
    else:
        msg = "high PM2.5, but it is breezy"
else:
    msg = "below the screening level"

print(msg)
```

<div style="border-left:7px solid #2f80ed;background:rgba(47,128,237,.12);padding:10px 18px;margin:10px 0;border-radius:8px;line-height:1.5">
    Does the test after <code>if</code> have to be a comparison? No. Python will take any value and ask whether it counts as true. Zero, an empty string and an empty list count as false. Nearly everything else counts as true, including the list <code>[0]</code>, which is not empty even though the only thing in it is a zero.
</div>

```python
print(bool(0), bool(1))
print(bool(""), bool("A"))
print(bool([]), bool([0]))
```

<div style="border-left:7px solid #2f80ed;background:rgba(47,128,237,.12);padding:10px 18px;margin:10px 0;border-radius:8px;line-height:1.5">
    How do you avoid rewriting the same <code>if</code>, <code>elif</code>, <code>else</code> every time you want a band? Write it once and give it a name. Everything pushed in under <code>def band(pm25):</code> is the body of a <b>function</b>, and <code>pm25</code> holds whatever value I hand over when I call it. <code>return</code> sends the answer back. After that, <code>band(41.2)</code> has the same shape as <code>type(7)</code>, and I can call it again with 35.0, or with 8.
</div>

```python
def band(pm25):
    if pm25 < 12:
        return "low"
    elif pm25 < 35:
        return "moderate"
    else:
        return "high"        # the same invented bands as before

print(band(41.2))
print(band(35.0))
print(band(8))
```

<div style="border-left:7px solid #2f80ed;background:rgba(47,128,237,.12);padding:10px 18px;margin:10px 0;border-radius:8px;line-height:1.5">
    What is the difference between a function that prints its answer and one that returns it? Take two ways to double a number. The first, <code>show_double(3)</code>, prints 6 on the screen. The second, <code>double(3)</code>, prints nothing, but the 6 comes back to the line that called it, where I can store it. Store the result of the first and you get <code>None</code>, Python's word for "nothing", because the function never returned anything.
</div>

```python
def show_double(x):
    print(x * 2)

def double(x):
    return x * 2

a = show_double(3)      # prints 6 on the way
b = double(3)           # prints nothing
print(a, b)
```

<div style="border-left:7px solid #2f80ed;background:rgba(47,128,237,.12);padding:10px 18px;margin:10px 0;border-radius:8px;line-height:1.5">
    Can a function have a setting that you usually leave alone? Give the value a default. In <code>def exceeds(reading, limit=35):</code>, a call with one value, <code>exceeds(41.2)</code>, uses the limit of 35. A call that names the setting, <code>exceeds(41.2, limit=50)</code>, replaces it. Writing <code>limit=50</code> in a call is a <b>keyword argument</b>: the name tells Python which setting you mean, so you do not have to remember the order.
</div>

```python
def exceeds(reading, limit=35):      # 35 is an invented limit
    return reading > limit

print(exceeds(41.2))
print(exceeds(41.2, limit=50))
print(exceeds(41.2, 50))             # the same, by position
```

<div style="background:linear-gradient(90deg,#2f80ed,#56ccf2);color:white;padding:10px 20px;border-radius:10px;margin:22px 0 6px 0">
    <span style="font-size:1.3em;font-weight:800">Section 5</span><span style="font-size:1.2em;font-weight:600"> &nbsp;Loops and Generators</span>
    <div style="opacity:.92">
        <code>for</code> and <code>while</code>, <code>break</code> and <code>continue</code>, and <code>yield</code>
    </div>
</div>

<div style="border-left:7px solid #2f80ed;background:rgba(47,128,237,.12);padding:10px 18px;margin:10px 0;border-radius:8px;line-height:1.5">
    <div style="font-weight:700;font-size:1.08em;margin-bottom:4px">
        💡 Two ways to repeat
    </div>
    <ul>
        <li>
            <code>for item in a_sequence:</code> runs the indented lines <b>once for each item</b> in a list, a string, a dictionary or a <code>range()</code>.
        </li>
        <li>
            <code>while condition:</code> repeats <b>as long as the condition stays True</b>.
        </li>
        <li>
            <code>break</code> leaves the loop at once. <code>continue</code> skips the rest of <b>this pass</b> and moves on to the next.
        </li>
        <li>
            <code>range(a, b, step)</code> counts from <code>a</code> up to, but <b>not including</b>, <code>b</code>. Same “stop is the first one you do not want” rule as slicing.
        </li>
    </ul>
</div>

<div style="border-left:7px solid #2f80ed;background:rgba(47,128,237,.12);padding:10px 18px;margin:10px 0;border-radius:8px;line-height:1.5">
    How do you do the same thing to every site in a list? You write it once and let Python repeat it. <code>for site in ["A", "B", "C"]:</code> runs the indented lines once for each item, and on each pass the name <code>site</code> holds the next one: first <code>"A"</code>, then <code>"B"</code>, then <code>"C"</code>. The same works on the characters of a string, and on <code>range(3)</code>, which hands out 0, 1 and 2 and stops before 3.
</div>

```python
for site in ["A", "B", "C"]:
    print("site", site)

for ch in "PM25":
    print(ch)

for i in range(3):
    print("pass", i)

print(list(range(10, 0, -1)))   # a third number is the step; a negative step counts down
```

<div style="border-left:7px solid #2f80ed;background:rgba(47,128,237,.12);padding:10px 18px;margin:10px 0;border-radius:8px;line-height:1.5">
    What is <code>count</code> when this loop ends? Start with <code>count = 0</code>, then <code>while count &lt; 9: count += 2</code>. Work it through. Python checks the condition before every pass. At 0, 2, 4, 6 and 8 the condition is True, so each earns another pass, and the pass at 8 makes it 10. Is 10 below 9? No, so the loop stops with <code>count</code> equal to 10, not 9. If nothing inside ever made the condition False, the cell would run forever, and you would stop it with Interrupt in the Kernel menu (Runtime, in Colab).
</div>

```python
count = 0
while count < 9:
    count += 2
print(count)
```

<div style="border-left:7px solid #2f80ed;background:rgba(47,128,237,.12);padding:10px 18px;margin:10px 0;border-radius:8px;line-height:1.5">
    How do you skip a reading you do not want? Suppose the readings are <code>[5, -9999, 7]</code> and -9999 means missing. Add them all and you get -9987, which is no concentration anyone measured. Test each one instead: if <code>x &lt; 0</code>, write <code>continue</code>, and Python abandons that pass and goes straight to the next. Only 5 and 7 reach <code>total += x</code>, and the total is 12.
</div>

```python
total = 0
for x in [5, -9999, 7]:
    if x < 0:
        continue
    total += x
print(total)
```

<div style="border-left:7px solid #2f80ed;background:rgba(47,128,237,.12);padding:10px 18px;margin:10px 0;border-radius:8px;line-height:1.5">
    How do you stop looking once you have found what you want? Take the readings <code>[38, 52, 61, 74, 69, 70, 71]</code> and ask for the first one above 70. Loop through them, and when <code>x &gt; 70</code> is True, store <code>x</code> and write <code>break</code>. The loop ends there, at 74. Without the <code>break</code> it would carry on, each new match would overwrite the last, and the answer would be 71.
</div>

```python
episode = [38, 52, 61, 74, 69, 70, 71]
first_over_70 = None

for x in episode:
    if x > 70:
        first_over_70 = x
        break

print(first_over_70)
```

<div style="border-left:7px solid #2f80ed;background:rgba(47,128,237,.12);padding:10px 18px;margin:10px 0;border-radius:8px;line-height:1.5">
    What does this script count? Read it before you run it. There are readings for three sites, a limit of 40, and a loop inside a loop: for each site, count the readings over the limit and file the count under the site's name. So <code>report</code> ends up a dictionary with the same three keys, and each value is a count. For B, the readings are 22, 35, 41 and 30, and only 41 is over 40, so I expect <code>report["B"]</code> to be 1.
</div>

```python
readings = {"A": [38, 52, 61, 74], "B": [22, 35, 41, 30], "C": [12, 18, 15, 20]}
limit = 40
report = {}

for site, values in readings.items():
    n_over = 0
    for x in values:
        if x > limit:
            n_over += 1
    report[site] = n_over

print(report)
```

<div style="border-left:7px solid #2f80ed;background:rgba(47,128,237,.12);padding:10px 18px;margin:10px 0;border-radius:8px;line-height:1.5">
    Why is <code>n_over = 0</code> inside the outer loop and not above it? Suppose I move it above. Then it is set to zero once and never again, so the three readings over the limit at A are still in the count when the loop reaches B, and B's one is added on top. The report comes out as <code>{"A": 3, "B": 4, "C": 4}</code>. It looks plausible, and it is wrong. When a counter lives inside a loop, ask when it resets.
</div>

```python
n_over = 0                      # moved above the loop: this is the mistake
wrong = {}

for site, values in readings.items():
    for x in values:
        if x > limit:
            n_over += 1
    wrong[site] = n_over

print(wrong)
```

<div style="border-left:7px solid #2f80ed;background:rgba(47,128,237,.12);padding:10px 18px;margin:10px 0;border-radius:8px;line-height:1.5">
    How do you reuse the same loop to ask a different question? Keep its shape and change what goes inside. For each site's highest reading, start with an empty dictionary and store <code>max(values)</code> under the site's name. For the sites that went over the limit at least once, start with an empty list and <code>.append(site)</code> whenever the count is above zero.
</div>

```python
peaks = {}
for site, values in readings.items():
    peaks[site] = max(values)
print(peaks)

flagged = []
for site, n in report.items():
    if n > 0:
        flagged.append(site)
print(flagged)
```

<div style="border-left:7px solid #2f80ed;background:rgba(47,128,237,.12);padding:10px 18px;margin:10px 0;border-radius:8px;line-height:1.5">
    How can a loop get its values without all of them existing at once? Put <code>yield</code> where a function would put <code>return</code>. Call <code>countdown(3)</code> and nothing runs yet: Python hands back a <b>generator</b>. Each time a loop asks for the next value, the function runs until it reaches <code>yield</code>, hands that value over, and waits. The next ask resumes on the line after the <code>yield</code>, with <code>n</code> as it was left. So the loop sees 3, then 2, then 1.
</div>

```python
def countdown(n):
    while n > 0:
        yield n
        n -= 1

for x in countdown(3):
    print(x)
```

<div style="border-left:7px solid #2f80ed;background:rgba(47,128,237,.12);padding:10px 18px;margin:10px 0;border-radius:8px;line-height:1.5">
    Can you loop over a generator twice? Ask for its values by hand and watch. <code>next(gen)</code> gives 3, and again gives 2. Collect what is left with <code>list(gen)</code> and you get <code>[1]</code>. Collect again and you get <code>[]</code>. A generator hands out each value once and is then spent, so to go through the values a second time you call <code>countdown(3)</code> again.
</div>

```python
gen = countdown(3)

print(next(gen))
print(next(gen))
print(list(gen))
print(list(gen))
```

<div style="border-left:7px solid #2f80ed;background:rgba(47,128,237,.12);padding:10px 18px;margin:10px 0;border-radius:8px;line-height:1.5">
    Do you need a whole function every time? No. Put a loop and a test inside parentheses and you have a generator in one line. To count the readings over 40 in <code>[38, 52, 61, 74]</code>, write <code>sum(1 for x in values if x &gt; 40)</code>. For each reading over 40 the generator hands out a 1, and <code>sum</code> adds them up as they arrive. The answer is 3, the same as the loop inside a loop gave for site A.
</div>

```python
values = [38, 52, 61, 74]

print(sum(1 for x in values if x > 40))
print(type(x for x in values))
```
