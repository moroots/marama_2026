## From plot to compliance question

The plot shows several high-PM2.5 hours in early June. But air quality standards are not set on single hours. The **24-hour PM2.5 standard** compares a *daily average* to a limit. So the next question is:

> **Did any site have a day that exceeded the 24-hour standard of 35 µg/m³?**

Before we can answer, we need a *daily mean per site*, and a rule for when a day has enough data to count.

### Step 1. Daily mean per site, with a completeness rule

For each site and each day, compute the mean **and** count how many valid hourly readings went into it. A day counts only if it has at least **18 of 24 hours** (75%).

```python
df["Day"] = df["Datetime"].dt.date

daily = (
    df.groupby(["Site ID", "Day"])["Sample Value"]
    .agg(daily_mean="mean", valid_hours="count")   # count() skips missing values
    .reset_index()
)

daily["complete"] = daily["valid_hours"] >= 18

print(f"{len(daily)} site-days in total")
print(f"{daily['complete'].sum()} have at least 18 valid hours ({daily['complete'].mean():.0%})")

daily_valid = daily[daily["complete"]]
daily_valid.head()
```

**Read and predict:** the missing hourly readings (the ones with a *Null Data Code*) are skipped when we count. Why does a missing hour not count as zero?

### Step 2. Compare each site-day to the 24-hour standard (35 µg/m³)

```python
NAAQS_24HR = 35   # µg/m³, primary 24-hour PM2.5 standard

exceed = daily_valid[daily_valid["daily_mean"] > NAAQS_24HR]
print("Site-days above 35 µg/m³:", len(exceed))
print("Highest HOURLY value in the file:   ", df["Sample Value"].max(), "µg/m³")
print("Highest valid DAILY mean in the file:", round(daily_valid["daily_mean"].max(), 1), "µg/m³")
```

Notice the gap. Individual *hours* reached well above 35, but no *day* did. The averaging time matters. The plot below shows every site's daily mean against the standard.

```python
fig, ax = plt.subplots(figsize=(10, 5), layout="tight")

for site, grp in daily_valid.groupby("Site ID"):
    ax.plot(pd.to_datetime(grp["Day"]), grp["daily_mean"], label=site, linewidth=1)

ax.axhline(NAAQS_24HR, color="black", linestyle="--", linewidth=2, label="24-hr standard (35)")
ax.set_ylim(0, 40)
ax.set_ylabel(r"24-hour mean ($\mu\mathrm{g}/m^3$)")
ax.set_xlabel("Date")
ax.set_title("Daily mean PM2.5 by site vs. the 24-hour standard")
ax.grid("both")
ax.legend(title="Site IDs", bbox_to_anchor=(1.01, 1), loc="upper left")
plt.show()
```

### Step 3. Compare sites side by side

The regulatory statistic for the 24-hour standard is the **98th percentile** of daily means (averaged over three years). With one year of data we can only *approximate* it. Treat this as a screening number, not a design value.

```python
site_summary = (
    daily_valid.groupby("Site ID")["daily_mean"]
    .agg(valid_days="count", p98=lambda s: s.quantile(0.98), max_daily="max")
    .round(1)
    .sort_values("p98", ascending=False)
)
site_summary
```

```python
fig, ax = plt.subplots(figsize=(8, 4), layout="tight")
ax.bar(site_summary.index, site_summary["p98"], color="tab:blue")
ax.axhline(NAAQS_24HR, color="black", linestyle="--", linewidth=2, label="24-hr standard (35)")
ax.set_ylabel(r"98th percentile of daily means ($\mu\mathrm{g}/m^3$)")
ax.set_xlabel("Site ID")
ax.set_title("Approximate 98th percentile by site, 2025")
ax.legend()
plt.show()
```

**Be careful with sites 0067 and 1008.** They have far fewer valid days than the others (see the `valid_days` column), so their numbers are not comparable. That is a data completeness problem, and it matters for regulatory use.

### Step 4. Annual mean vs. the annual standard (9.0 µg/m³)

The annual PM2.5 standard is **9.0 µg/m³**. A simple calendar-year mean of valid daily means gives a first look. The official value uses quarterly averaging, three years of data and strict completeness rules, so this is a screening estimate only.

```python
NAAQS_ANNUAL = 9.0   # µg/m³, primary annual PM2.5 standard

annual = (
    daily_valid.groupby("Site ID")["daily_mean"]
    .agg(annual_mean="mean", valid_days="count")
    .round(1)
    .sort_values("annual_mean", ascending=False)
)
annual["above_9"] = annual["annual_mean"] > NAAQS_ANNUAL
annual
```

```python
fig, ax = plt.subplots(figsize=(8, 4), layout="tight")
colors = ["tab:red" if v else "tab:blue" for v in annual["above_9"]]
ax.bar(annual.index, annual["annual_mean"], color=colors)
ax.axhline(NAAQS_ANNUAL, color="black", linestyle="--", linewidth=2, label="Annual standard (9.0)")
ax.set_ylabel(r"Annual mean ($\mu\mathrm{g}/m^3$)")
ax.set_xlabel("Site ID")
ax.set_title("Screening annual mean PM2.5 by site, 2025")
ax.legend()
plt.show()
```

### What did we learn?

- The early-June spike is real, but the **24-hour standard** was not exceeded at any site in this file.
- The **annual** comparison tells a different story. The result depends on which standard you ask about.
- **Averaging time, completeness and the choice of statistic** change the answer. Always ask what number a standard actually compares.
- This is a screening look at one year. Official determinations use three years and specific rules.
